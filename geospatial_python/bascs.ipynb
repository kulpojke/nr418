### A shorter way: list comprehensions

A **list comprehension** builds a new list in one line. It combines the loop and `.append()` operation:

```python
[expression for item in collection]
```

Read it as: “For each item in the collection, calculate the expression and put the result in a new list.” Add an optional `if` at the end to keep only items that meet a condition.

Here are the two loops from Part 5 written as comprehensions. The first creates a boolean for every score; the second keeps only scores above the threshold.


````python
predictions = [score > threshold for score in scores]
flagged_scores = [score for score in scores if score > threshold]
print(predictions)
print(flagged_scores)
````

**Expected output:**

